```python
import cv2
import numpy as np
from matplotlib import pyplot as plt
plt.rcParams['figure.figsize'] = [15, 10]
```

```python
img = cv2.imread('5719cdddaafd41543ccc2a04.jpg')
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
plt.imshow(img_rgb)
```

```python
red, green, blue = cv2.split(img_rgb)
img_rgb1 = cv2.merge([red, green, blue])
img_rgb2 = cv2.merge([red, blue, green])
img_rgb3 = cv2.merge([green, red, blue])
img_rgb4 = cv2.merge([blue, green, red])
out1 = np.hstack([img_rgb1, img_rgb2])
out2 = np.hstack([img_rgb3, img_rgb4])
out = np.vstack([out1, out2])
plt.imshow(out)
plt.axis(False)
```

```python
img = cv2.imread('5719cdddaafd41543ccc2a04.jpg')
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB) 

img_lr = np.fliplr(img)
out1 = np.hstack([img,img_lr])
out2 = np.flipud(out1)
out = np.vstack([out1, out2])

plt.imshow(out)
plt.axis('off')
```
